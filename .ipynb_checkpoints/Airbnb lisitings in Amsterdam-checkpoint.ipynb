```python
import os 
import pandas as pd
import pandas as pd
from pandas.api.types import is_string_dtype
from pandas.api.types import is_numeric_dtype
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import csv
#data_direct = os.path.join('..','AirBNB','data')
```

```python
#read listing data

listings_data = pd.read_csv('listings.csv')
```

```python
#print data types 

listings_data.dtypes
```

```python
# print listings info
listings_data.info
```

```python
listings_data.columns
```

```python
# reduce to relevant columns

listings = listings_data[['id','neighbourhood','neighbourhood_cleansed','availability_60','availability_365','availability_30','latitude','longitude','property_type','room_type','accommodates','price','host_since',
    'guests_included','minimum_nights','number_of_reviews','review_scores_rating','instant_bookable','cancellation_policy']]
```

```python
# correlation between relevant variables
corr = listings.corr(method='kendall')
plt.figure(figsize=(15,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
listings.columns
```

```python
listings.head()
```

## Predicting Prices 

## Listing count by type

```python
sns.countplot(listings['room_type'], palette="coolwarm")
fig = plt.gcf()
fig.set_size_inches(20,10)
plt.title('Room Type')
```

## Listing availbility by type

```python
plt.figure(figsize=(10,10))
ax = sns.boxplot(data= listings, x='room_type',y='availability_365',palette='coolwarm')
```
